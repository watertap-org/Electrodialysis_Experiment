# Selective Electrodialysis (SED) Simulation Assisted by a Data-driven Model Component and Available Experiment Data 

## Background
The key to excelling a SED model is to effectively estimate or predict the cation transport number across cation exchange membranes (CEMs). This cross-membrane transport number for ion i, denoted as ti’, controls the treatment efficiency of a specific ion and therefore ion selectivity of the entire process.    Direct experimental measurement of ti’ in SED is difficult and hardly captures its variation spatially or temporally that is present in practical processes due to its dependence on ion concentrations.
